```python
import pandas as pd

# Ganti jalur di bawah ini jika posisinya berbeda di Drive Anda
file_path = '/content/drive/My Drive/ecomerce sales data/Project_01_Ecommerce_Sales_Raw_Data.xlsx'

# Membaca file excel
df = pd.read_excel(file_path)

# Menampilkan 5 baris pertama data
df.head()
```

```python
# Membaca file Excel langsung dari Drive
df_raw = pd.read_excel(file_path)

# Menampilkan 5 baris pertama
display(df_raw.head())
```

```python
# 3. CREATE WORKING COPY
# ==========================================

# Menyimpan salinan dataset asli
df_work = df_raw.copy()

# Menambahkan nomor baris sumber untuk pelacakan
df_work['Source_Row'] = df_work.index

print("Raw data rows:", len(df_raw))
print("Working data rows:", len(df_work))
```

```python
# ==========================================
# 4. INITIAL DATA INSPECTION
# ==========================================

# Melihat ukuran dataset: (baris, kolom)
print("Dataset shape:", df_work.shape)

# Melihat nama kolom
print("\nColumn names:")
print(df_work.columns.tolist())

# Melihat tipe data setiap kolom
print("\nData types:")
print(df_work.dtypes)

# Melihat 5 baris pertama
display(df_work.head())

# Ringkasan statistik
display(df_work.describe(include='all').T)
```

```python
# ==========================================
# 5. CHECK MISSING VALUES
# ==========================================

missing_report = pd.DataFrame({
    'Missing_Count': df_work.isnull().sum(),
    'Missing_Percentage': (
        df_work.isnull().sum() / len(df_work) * 100
    )
})

display(missing_report)
```

```python
# Menampilkan semua baris yang memiliki minimal 1 nilai kosong

missing_rows = df_work[df_work.isnull().any(axis=1)]

print("Jumlah baris yang memiliki missing values:",
      len(missing_rows))

display(missing_rows)
```

```python
# ==========================================
# 6. CHECK DUPLICATE RECORDS
# ==========================================

# Menghitung seluruh baris duplikat
duplicate_count = df_work.duplicated().sum()

print("Jumlah duplicate rows:", duplicate_count)

# Menampilkan baris duplikat
duplicates = df_work[
    df_work.duplicated(keep=False)
]

display(duplicates)
```

```python
# Menghapus baris yang identik
# Source_Row tidak ikut digunakan dalam pemeriksaan

original_columns = df_raw.columns.tolist()

df_work = df_work.drop_duplicates(
    subset=original_columns,
    keep='first'
).copy()

print("Rows after removing duplicates:", len(df_work))
```

```python
# ==========================================
# 7. STANDARDIZE TEXT COLUMNS
# ==========================================

text_columns = [
    'Category',
    'Region',
    'Sales_Channel',
    'Order_Status'
]

for column in text_columns:
    df_work[column] = (
        df_work[column]
        .str.strip()
        .str.title()
    )

print("Text columns successfully standardized!")

display(df_work[text_columns].head())
```

```python
# Membersihkan nama produk

canonical_products = [
    "Wireless Mouse",
    "Mechanical Keyboard",
    "USB-C Hub",
    "Laptop Stand",
    "27-inch Monitor",
    "Webcam HD",
    "Bluetooth Speaker",
    "Desk Lamp",
    "Office Chair",
    "Notebook Set",
    "Pen Pack",
    "Desk Organizer"
]

product_lookup = {
    product.casefold(): product
    for product in canonical_products
}

df_work['Product'] = (
    df_work['Product']
    .str.strip()
    .str.casefold()
    .map(product_lookup)
)

display(df_work['Product'].value_counts(dropna=False))
```

```python
# ==========================================
# 8. STANDARDIZE DATE FORMAT
# ==========================================

def parse_date(value):

    if pd.isna(value):
        return pd.NaT

    # Jika sudah berbentuk datetime
    if isinstance(value, (pd.Timestamp,)):
        return value

    formats = [
        '%Y-%m-%d',
        '%m/%d/%Y',
        '%Y/%m/%d',
        '%d-%m-%Y'
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(value, format=fmt)
        except (ValueError, TypeError):
            continue

    return pd.NaT


df_work['Order_Date'] = (
    df_work['Order_Date'].apply(parse_date)
)

print("Date cleaning completed!")

display(df_work[['Order_ID', 'Order_Date']].head(10))
print("\nMissing dates:", df_work['Order_Date'].isna().sum())
```

```python
# ==========================================
# 9. CHECK INVALID NUMERIC VALUES
# ==========================================

# Memastikan kolom angka memiliki tipe numerik
df_work['Quantity'] = pd.to_numeric(
    df_work['Quantity'],
    errors='coerce'
)

df_work['Unit_Price'] = pd.to_numeric(
    df_work['Unit_Price'],
    errors='coerce'
)

# Membuat tanda untuk data yang bermasalah
df_work['Invalid_Quantity'] = (
    df_work['Quantity'].isna() |
    (df_work['Quantity'] <= 0)
)

df_work['Invalid_Unit_Price'] = (
    df_work['Unit_Price'].isna() |
    (df_work['Unit_Price'] <= 0)
)

# Menampilkan data bermasalah
invalid_rows = df_work[
    df_work['Invalid_Quantity'] |
    df_work['Invalid_Unit_Price']
]

display(invalid_rows[
    ['Order_ID', 'Quantity', 'Unit_Price',
     'Invalid_Quantity', 'Invalid_Unit_Price']
])
```

```python
# ==========================================
# 10. CHECK CATEGORICAL VALUES
# ==========================================

columns_to_check = [
    'Category',
    'Region',
    'Sales_Channel',
    'Order_Status'
]

for column in columns_to_check:

    print("\nColumn:", column)

    display(
        df_work[column].value_counts(dropna=False)
    )
```

```python
# ==========================================
# 11. DATA CLEANING AUDIT REPORT
# ==========================================

audit_report = pd.DataFrame({
    'Metric': [
        'Original Rows',
        'Rows After Duplicate Removal',
        'Duplicate Rows Removed',
        'Missing Order Date',
        'Missing Product',
        'Missing Quantity',
        'Missing Unit Price',
        'Invalid Quantity (Missing or <= 0)',
        'Invalid Unit Price (Missing or <= 0)',
        'Missing Region',
        'Missing Sales Channel'
    ],

    'Value': [
        len(df_raw),
        len(df_work),
        len(df_raw) - len(df_work),
        df_work['Order_Date'].isna().sum(),
        df_work['Product'].isna().sum(),
        df_work['Quantity'].isna().sum(),
        df_work['Unit_Price'].isna().sum(),
        df_work['Invalid_Quantity'].sum(),
        df_work['Invalid_Unit_Price'].sum(),
        df_work['Region'].isna().sum(),
        df_work['Sales_Channel'].isna().sum()
    ]
})

display(audit_report)
```

```python
# ==========================================
# 11. DATA CLEANING AUDIT REPORT
# ==========================================

audit_report = pd.DataFrame({
    'Metric': [
        'Original Rows',
        'Rows After Duplicate Removal',
        'Duplicate Rows Removed',
        'Missing Order Date',
        'Missing Product',
        'Missing Quantity',
        'Missing Unit Price',
        'Invalid Quantity (Missing or <= 0)',
        'Invalid Unit Price (Missing or <= 0)',
        'Missing Region',
        'Missing Sales Channel'
    ],

    'Value': [
        len(df_raw),
        len(df_work),
        len(df_raw) - len(df_work),
        df_work['Order_Date'].isna().sum(),
        df_work['Product'].isna().sum(),
        df_work['Quantity'].isna().sum(),
        df_work['Unit_Price'].isna().sum(),
        df_work['Invalid_Quantity'].sum(),
        df_work['Invalid_Unit_Price'].sum(),
        df_work['Region'].isna().sum(),
        df_work['Sales_Channel'].isna().sum()
    ]
})

display(audit_report)
```

```python
# ==========================================
# 12. CREATE ANALYSIS DATASET
# ==========================================

# Hanya mengambil pesanan Completed
df_analysis = df_work[
    df_work['Order_Status'] == 'Completed'
].copy()

# Memisahkan baris yang bisa dan belum bisa
# digunakan untuk menghitung revenue

valid_revenue_mask = (
    df_analysis['Quantity'].notna() &
    (df_analysis['Quantity'] > 0) &
    df_analysis['Unit_Price'].notna() &
    (df_analysis['Unit_Price'] > 0)
)

df_revenue_excluded = df_analysis[
    ~valid_revenue_mask
].copy()

df_analysis = df_analysis[
    valid_revenue_mask
].copy()

print("Completed orders:", len(
    df_work[df_work['Order_Status'] == 'Completed']
))

print("Valid revenue records:", len(df_analysis))

print("Completed records excluded from revenue:",
      len(df_revenue_excluded))
```

```python
# ==========================================
# 13. CALCULATE REVENUE
# ==========================================

df_analysis['Revenue'] = (
    df_analysis['Quantity'] *
    df_analysis['Unit_Price'] *
    (1 - df_analysis['Discount'])
)

display(
    df_analysis[
        ['Order_ID', 'Quantity', 'Unit_Price',
         'Discount', 'Revenue']
    ].head(10)
)

print("Total Revenue: £",
      round(df_analysis['Revenue'].sum(), 2))
```

```python
# ==========================================
# 14. EXPORT CLEANED DATA
# ==========================================

# Menyimpan dataset hasil cleaning
df_work.to_csv(
    'ecommerce_cleaned_all_records.csv',
    index=False
)

# Menyimpan dataset khusus perhitungan revenue
df_analysis.to_csv(
    'ecommerce_valid_revenue_data.csv',
    index=False
)

# Menyimpan laporan audit
audit_report.to_csv(
    'data_cleaning_audit_report.csv',
    index=False
)

print("All files successfully exported!")
```

```python
from datetime import datetime, date

def parse_date(value):

    if pd.isna(value):
        return pd.NaT

    if isinstance(value, (pd.Timestamp, datetime, date)):
        return pd.to_datetime(value)

    formats = [
        '%Y-%m-%d',
        '%m/%d/%Y',
        '%Y/%m/%d',
        '%d-%m-%Y'
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(value, format=fmt)
        except (ValueError, TypeError):
            continue

    return pd.NaT
```
